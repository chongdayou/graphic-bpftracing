```python
import csv
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict
from matplotlib.lines import Line2D
from PIL import Image
```

```python
cpu_count = '1'
file_path = 'trace_mutex_latency_' + cpu_count + '_CPU_results.csv'
```

```python
stats = defaultdict(list)
with open(file_path, 'r') as file:
    ms_past = 0
    for line in file:
        if 'Averages' in line:
            ms_past = int(line[line.find('@')+1:line.find('ms')])
        if line[:19] == '@mutex_latency_avgs':
            tid = int(line[20:line.find(']')])
            avg = int(line[line.find(':')+2:])
            avg_ms = avg / 1000000.0
            stats[tid].append( (ms_past,avg_ms) )
            
```

```python
print(stats)
```

```python
plt.figure(figsize=(10,5))
plt.xlabel('ms past since process starts')
plt.ylabel('avg ms mutex hold latency')

for tid, points in stats.items():
    x_values = [p[0] for p in points]
    y_values = [p[1] for p in points]
    plt.plot(x_values, y_values, alpha=0.8, lw=2)

plt.xticks(np.arange(90, 261, 10))

plt.savefig('plots/latency_avg_over_time_' + cpu_count + '_CPU.png', dpi=300)
plt.close()
```
